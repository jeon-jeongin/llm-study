```python
""" precision & recall (분석성능평가지표)
TP(True Positive) : 관련있고 검색도 됨(맞게 포착)
FP(False Positive) : 무관한데 검색에 나옴(노이즈)
FN(False Negative) : 관련있는데 검색 못 함(누락)
TN(True Negative) : 무관하고 검색도 안 됨(올바른 제외)

precision : 정밀도 - 얼마나 정밀한지 확인 TP/(TP+FP)
    검색된 것 중 진짜 관련인 비율
    Precision@K = (상위 K개 중 관련 문서 수) / K
    K가 커질수룩 관련 없는 문서가 섞여 들어와 Precision은 하락하는 경향

    보여준 것 중 쓸모 있는 건 몇 %?
    모르는 것 : 놓친 관련 문서 존재 여부
    낮으면 LLM에 노이즈 청크 전달 -> 환각 유발 

recall : 재현율 TP/(TP+FN)
    전체 관련 문서 중 잡아낸 비율
    Recall@K = (상위 K개 중 관련 문서 수) / 전체 관련 문서 수(R)
    K가 커질수룩 관련 문서를 더 많이 포함하므로 Recall은 상승하는 경향

    있는 것 중 못 찾은 건 없나?
    모르는 것 : 결과에 노이즈가 얼마나 섞였는지
    핵심 정보 누락 -> 불완전한 답변 생성

Trade-off가 발생
검색 시스템은 관련도 점수(relevance score) 기준으로 문서를 정렬한 뒤, 임계값 이상의 문서만 반환
K 또는 임계 값을 연속적으로 변화시키며(Recallm Precision) 점을 이어 PR 곡선을 만듬
곡선이 (1,1)에 가까울수룩 좋은 시스템
AP(Average Precision) : PR 곡선 아래 면적 = 종합 성능 지표
이상적 시스템 : Recall=1&Precision=1

R-Precision = 상위 R개 중 관련 문서 수 / R
Precision@K의 한계 : K를 임의로 정해야 함. 쿼리마다 관련 문서 수가 다르면 불공표한 비교
R-Precision의 장점 : 각 쿼리의 관련 문서 수 R을 자동으로 K로 사용 -> "같은 조건"에서 공정하게 비교

R-Precision에서 Precision = Recall
R-Precision = TP / R = TP/(TP+FN) = Recall@R
수식적으로는 같지만 - 분모를 K로 보는가, R로 보는가에 대한 관점 차이

F1@K : 조화평균 2 * P@K * R@K / (P@K + R@K)
    둘 중 하나라도 낮으면 전체 점수가 크게 떨어지도록 패널티 부여
F-beta : 중요도 조절
"""
```

```python
langchain_docs = [
    Document(page_content=doc["content"], metadata={"doc_id": doc["doc_id"], "title": doc["title"]})
    for doc in documents
]
vectorstore = FAISS.from_documents(langchain_docs, embeddings_model)
def search_documents(query: str, k: int = 5) -> List[Dict]:
    results = vectorstore.similarity_search_with_score(query, k=k)
    retrieved = []
    for doc, score in results:
        retrieved.append({
            "doc_id": doc.metadata["doc_id"],
            "title": doc.metadata["title"],
            "content": doc.page_content,
            "score": float(score)
        })
    return retrieved
search_results_cache = {}
for qa in qa_dataset:
    results = search_documents(qa["question"], k=5)
    search_results_cache[qa["query_id"]] = results
```

```python
def precision_at_k(query_id, k):
    qa = next(q for q in qa_dataset if q['query_id'] == query_id)
    relevant_ids = set(qa['relevant_doc_ids'])
    retrieved = search_results_cache[query_id][:k]
    retrieved_ids = {r['doc_id'] for r in retrieved}

    relevant_retrieved = relevant_ids & retrieved_ids
    return len(relevant_retrieved) / k
```

```python
def recall_at_k(query_id, k):
    qa = next(q for q in qa_dataset if q['query_id'] == query_id)
    relevant_ids = set(qa['relevant_doc_ids'])
    retrieved = search_results_cache[query_id][:k]
    retrieved_ids = {r['doc_id'] for r in retrieved}

    relevant_retrieved = relevant_ids & retrieved_ids
    return len(relevant_retrieved) / len(relevant_ids) if relevant_ids else 0.0
```

```python
def f1_at_k(query_id, k):
    p = precision_at_k(query_id, k)
    r = recall_at_k(query_id, k)
    
    if p+r == 0:
        return 0.0
        
    return 2*p*r / (p+r)
```

```python
def r_precision(query_id):
    qa = next(q for q in qa_dataset if q['query_id'] == query_id)
    relevant_ids = set(qa['relevant_doc_ids'])
    R = len(relevant_ids)

    if R == 0:
        return 0.0

    return precision_at_k(query_id, R)
```

```python
def f_beta_at_k(query_id, k, beta):
    p = precision_at_k(query_id, k)
    r = recall_at_k(query_id, k)
    
    beta_sq = beta ** 2
    
    if beta_sq * p + r == 0:
        return 0.0
        
    return (1 + beta_sq) * p * r / (beta_sq * p + r)
```

```python
"""
전통 NLP 메트릭
Rag가 생성한 답변의 품질을 정답(ground_truth)과 직접 텍스트 비교로 평가하는 방법

BLEU-n = (생성 답변의 n-gram 중 정답에도 등장하는 수) / (생성 답변의 전체 n-gram 수)
BERTScore : 의미 기반 유사도
"""
```

```python
from collections import Counter
import numpy as np
```

```python
def get_ngrams(tokens, n):
    return Counter(tuple(tokens[i:i+n]) for i in range(len(tokens) - n + 1))
```

```python
def bleu_score(reference, hypothesis, max_n):

    ref_tokens = reference.split()
    hyp_tokens = hypothesis.split()

    c, r = len(hyp_tokens), len(ref_tokens)
    bp = 1.0 if c >- r else np.exp(1-r/c) if c > 0 else 0.0

    precisions = {}
    for n in range(1, max_n+1):
        ref_ngrams = get_ngrams(ref_tokens, n)
        hyp_ngrams = get_ngrams(hyp_tokens, n)

        if len(hyp_ngrams) == 0:
            precisions[0] = 0.0
            continue

        clipped = sum(min(cnt, ref_ngrams.get(ng, 0)) for ng, cnt in hyp_ngrams.items())
        precisions[n] = clipped / sum(hyp_ngrams.values())

    log_avg, valid_n = 0.0, 0
    
    for n in range(1, max_n + 1):
        if precisions[n] > 0:
            log_avg += np.log(precisions[n])
            valid_n += 1

    bleu = bp * np.exp(log_avg / valid_n) if valid_n > 0 else 0.0

    return {'bleu': round(bleu, 4), 'bp': round(bp, 4)}
```

```python
ref = "트랜스포머의 핵심 메커니즘은 셀프 어텐션으로, 입력 시퀀스의 모든 위치 간 관계를 병렬로 처리합니다."
hyp = "트랜스포머는 셀프 어텐션 메커니즘을 사용하여 시퀀스 내 모든 위치 간 관계를 병렬로 처리합니다."
```

```python
result = bleu_score(ref, hyp, 4)
```

```python
# 검색에서
def evaluate_context_relevance(question, contexts, model):
    """LLM을 사용하여 검색된 컨텍스트의 관련성 평가"""
    context_text = '\n\n'.join([
        f"[컨텍스트 {i+1}] : {ctx}" for i, ctx in enumerate(contexts)
    ])

    prompt = f"""질문과 검색된 컨텍스트의 관련성을 평가하세요.

    ## 질문
    {question}

    ## 검색된 컨텍스트
    {context_text}

    ## 평가기준
    각 컨텍스트에 대해 1~5점으로 평가하세요:
    - 5 : 질문에 직접적으로 답할 수 있는 핵심 정보 포함
    - 4 : 질문과 매우 관련 있는 정보 포함
    - 3 : 부분적으로 관련 있는 정보 포함
    - 2 : 간접적으로만 관련 있음
    - 1 : 거의 관련 없음

    ## 응답 형식(JSON)
    {{"scores" : [점수1, 점수2, ...], "average" : "평균점수", "reasoning" :"평가근거"}}

    반드시 유요한 JSON만 출력하세요.
    """

    response = model.bind(response_format = {'type' : 'json_object'}).invoke([HumanMessage(content=prompt)])
    result = json.loads(response.content)

    return result
```
