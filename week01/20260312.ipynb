```python
# temperature : 낮은 값에서는 동일한 값이 나오며 높아지면 다른 값들이 나오게 됨(모델의 창의성을 정의하는 파라미터)
```

```python
question = "인공지능의 미래에 대해 한 문장으로 말해주세요"
reponse = client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = [
        {'role' : 'user', 'content' : question},
    ],
    temperature = 1.5
)
```

```python
# top_p, top_k : 어떻게 sampling을 하는지
```

```python
question = "인공지능의 미래에 대해 한 문장으로 말해주세요"
settings = [
    (0.0, 30),
    (0.7, 60),
    (1.2, 100),
]
for temp, max_tok in settings:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": question}],
        temperature=temp,
        max_tokens=max_tok,
    )
    print(temp, max_tok, "=>", response.choices[0].message.content)
```

```python
# penalty
# frequency_penalty : 같은 단어가 계속 나오지 못하게 0 ~ 1 default=0, 0 : 반복 신경 안씀 / 1 : 반복하면 안됨
# presence_penalty : default=0 0: 기존 단어 나와도 신경 안씀 / 1 : 적극적으로 새로운 주제나 단어
```

```python
messages = [
    {"role": "system", "content": "당신은 여행 플래너입니다. 한국어로 짧고 친절하게 답해주세요."}
]

def chat_once(client, messages, user_input):
    messages.append({"role": "user", "content": user_input})
    
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
    )
    assistant_reply = response.choices[0].message.content
    messages.append({"role": "assistant", "content": assistant_reply})
    return assistant_reply, messages

reply, messages = chat_once(client, messages, "5월에 일본 도쿄로 3박 4일 여행 가려고 하는데, 대략적인 일정 추천해줄래?")
print("1턴:", reply)
reply, messages = chat_once(client, messages, "맛집 위주로 하루 일정을 다시 짜 줄 수 있어?")
print("2턴:", reply)
reply, messages = chat_once(client, messages, "지금까지 말해준 일정 중에 비 오는 날에도 괜찮은 코스만 골라 줄래?")
print("3턴:", reply)
```

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "파이썬의 장점 3가지를 설명해줘"}],
    stream = True
)
full_answer = ""
for chunk in response:
    text = chunk.choices[0].delta.content
    if text:
        print(text, flush=True)
        full_answer += text
```
