```python
def make_prompt(question, language='한국어'):
    prompt = f"다음 질문에 {language}로 답변해주세요 : {question}"
    return prompt
```

```python
def is_question(question):
    if '?' in question:
        return True
    else:
        return False
```

```python
is_question("test")
```

```python
is_question("test?")
```
